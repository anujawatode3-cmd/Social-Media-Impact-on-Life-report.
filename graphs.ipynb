```python
import pandas as pd

csv_file = r"C:\Users\Anuja\Downloads\archive (1)\Social_media_impact_on_life.csv"

df = pd.read_csv(csv_file)

print("Dataset loaded successfully!")
print("Dataset shape:", df.shape)

df.head()
```

```python
import numpy as np

numeric_data = df.select_dtypes(include="number")

print("Mean values:")
print(numeric_data.mean().round(2))

print("\nMinimum values:")
print(numeric_data.min())

print("\nMaximum values:")
print(numeric_data.max())
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

plt.figure(figsize=(8, 5))

sns.histplot(data=df, x="Daily_Usage_Hours", bins=20, kde=True)

plt.title("Distribution of Daily Social Media Usage")
plt.xlabel("Daily Usage (Hours)")
plt.ylabel("Number of Students")

plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

plt.figure(figsize=(9, 5))

sns.countplot(
    data=df,
    x="Primary_Platform",
    order=df["Primary_Platform"].value_counts().index
)

plt.title("Most Popular Social Media Platforms")
plt.xlabel("Social Media Platform")
plt.ylabel("Number of Students")
plt.xticks(rotation=45)

plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(8, 5))

sns.scatterplot(
    data=df,
    x="Daily_Usage_Hours",
    y="Sleep_Duration_Hours"
)

plt.title("Daily Social Media Usage vs Sleep Duration")
plt.xlabel("Daily Social Media Usage (Hours)")
plt.ylabel("Sleep Duration (Hours)")

plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(8, 5))

sns.scatterplot(
    data=df,
    x="Daily_Usage_Hours",
    y="Academic_Performance_GPA"
)

plt.title("Social Media Usage vs Academic Performance")
plt.xlabel("Daily Usage (Hours)")
plt.ylabel("Academic Performance (GPA)")

plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(12, 8))

numeric_data = df.select_dtypes(include="number")
correlation = numeric_data.corr()

sns.heatmap(
    correlation,
    annot=True,
    cmap="coolwarm",
    fmt=".2f",
    linewidths=0.5
)

plt.title("Correlation Heatmap of Social Media Impact")
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(8, 5))

sns.countplot(
    data=df,
    x="Overall_Impact",
    order=df["Overall_Impact"].value_counts().index
)

plt.title("Overall Impact of Social Media")
plt.xlabel("Overall Impact")
plt.ylabel("Number of Records")
plt.xticks(rotation=30)

plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(8, 5))

sns.boxplot(
    data=df,
    x="Late_Night_Usage",
    y="Sleep_Duration_Hours"
)

plt.title("Late-Night Social Media Usage vs Sleep Duration")
plt.xlabel("Late-Night Usage")
plt.ylabel("Sleep Duration (Hours)")

plt.tight_layout()
plt.show()
```
